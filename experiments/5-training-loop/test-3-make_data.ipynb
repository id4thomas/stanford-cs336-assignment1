```python
import numpy as np
import os

# ===== Create fake train/val token sequences =====
# We'll just make random integers representing token IDs
vocab_size = 1000
train_tokens = np.random.randint(0, vocab_size, size=(10_000,), dtype=np.uint16)
val_tokens   = np.random.randint(0, vocab_size, size=(2_000,), dtype=np.uint16)

os.makedirs("data", exist_ok=True)
train_tokens.tofile("data/train.bin")
val_tokens.tofile("data/val.bin")

print("Dummy data saved to data/train.bin and data/val.bin")
```

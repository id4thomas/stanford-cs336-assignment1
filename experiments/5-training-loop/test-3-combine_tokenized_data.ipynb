## Load Tokenizer, Prepare Data

```python
import os
import sys

prj_path = os.path.dirname(os.path.dirname(os.getcwd()))
sys.path.append(os.path.join(prj_path, "cs336_basics"))
# os.chdir(src_path)
from tokenizer.bpe import BPETokenizer
```

```python
import tiktoken
ref_tokenizer = tiktoken.encoding_for_model('gpt-4o')
```

```python
import pickle

# Load tinystories trained tokenizer
vocab_path = os.path.join(prj_path, "results/tokenizer/tinystories-v10000/vocab.pkl")
merges_path = os.path.join(prj_path, "results/tokenizer/tinystories-v10000/merges.pkl")

with open(vocab_path, 'rb') as f:
    vocab = pickle.load(f)
    
with open(merges_path, 'rb') as f:
    merges = pickle.load(f)
```

```python
tokenizer = BPETokenizer(
    vocab=vocab,
    merges=merges,
    special_tokens=["<|endoftext|>"]
)
```

```python
# Test Tokenizer
test_string = "Hello how are you? 🙃"
encoded_ids = tokenizer.encode(test_string)
print("ENCODED:", len(encoded_ids), encoded_ids)

decoded_string = tokenizer.decode(encoded_ids)
print("DECODED:", decoded_string)
```

```python
encoded_ids = ref_tokenizer.encode(test_string)
print("ENCODED:", len(encoded_ids), encoded_ids)
```

## Test Saving Tokenized Data

```python
import numpy as np
import os

# ===== Create fake train/val token sequences =====
# We'll just make random integers representing token IDs
vocab_size = 1000
train_tokens = np.random.randint(0, vocab_size, size=(10_000,), dtype=np.uint16)
val_tokens   = np.random.randint(0, vocab_size, size=(2_000,), dtype=np.uint16)

os.makedirs("data", exist_ok=True)
train_tokens.tofile("data/train.bin")
val_tokens.tofile("data/val.bin")

print("Dummy data saved to data/train.bin and data/val.bin")
```

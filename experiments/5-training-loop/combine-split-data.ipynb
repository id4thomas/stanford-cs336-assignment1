```python
import os
import numpy as np
```

# TinyStories

```python
data_dir = "../../results/dataset/TinyStoriesV2-GPT4-train"

data = []
for i in range(8):
    with open(os.path.join(data_dir, f"tokens/chunk{i}.npy"), 'rb') as f:
        chunk = np.load(f)
    print("Chunk {}: {}".format(i, chunk.shape))
    data.append(chunk)

data = np.concatenate(data, axis=0)
print(data.shape)

with open(os.path.join(data_dir, "data.npy"), 'wb') as f:
    np.save(f, data)
```
